```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df = pd.read_csv("data/covid19_sg.csv")

# convert the data column
df["Date"] = pd.to_datetime(df["Date"])
df = df.sort_values("Date")

# convert to incidence, use Daily Confirmed from data
# set singapore population
population = 5686000

df["7_day_incidence"] = (
    df["Daily Confirmed"]
        .rolling(window=7)
        .sum()
        * 100000
        / population
)


#visualize 7 day incidence



plt.figure(figsize=(14,6))

plt.plot(df["Date"], df["7_day_incidence"], color="blue")

plt.title("7-Day Incidence")
plt.xlabel("Date")
plt.ylabel("Cases per 100,000 inhabitants")

plt.grid(True)

plt.show()

```

```python
# Avoid division by zero
cases = df["Daily Confirmed"].replace(0, np.nan)
hospitalized = df["Still Hospitalised"].replace(0, np.nan)

df["hospitalization_rate"] = (
    hospitalized / cases
) * 100

plt.figure(figsize=(12, 5))
plt.plot(df["Date"], df["hospitalization_rate"], color="orange")
plt.title("Hospitalization Rate")
plt.xlabel("Date")
plt.ylabel("Hospitalization Rate (%)")
plt.grid(True)
plt.show()
```

```python
# Calculate and visualize the percentage of hospitalized patients requiring ICU care
df["criticality"] = (
    df["Intensive Care Unit (ICU)"] / hospitalized
) * 100

plt.figure(figsize=(12, 5))
plt.plot(df["Date"], df["criticality"], color="red")
plt.title("Criticality")
plt.xlabel("Date")
plt.ylabel("Hospitalized (%)")
plt.grid(True)
plt.show()
```

```python
# Visualize vaccination progress over time



plt.figure(figsize=(14,6))

plt.plot(
    df["Date"],
    df["Cumulative Individuals Vaccination Completed"],
    color="green",
    linewidth=2
)

plt.title("Cumulative Individuals Fully Vaccinated")
plt.xlabel("Date")
plt.ylabel("Number of Individuals")

plt.grid(alpha=0.3)
plt.tight_layout()

plt.show()
```

```python
# Death Rate
df["death_rate"] = (
    df["Daily Deaths"] /
    df["Daily Confirmed"].replace(0,np.nan)
) * 100

plt.figure(figsize=(14,6))

plt.plot(
    df["Date"],
    df["death_rate"],
    color="black"
)

plt.title("Daily COVID Death Rate")
plt.ylabel("Death Rate (%)")

plt.grid(alpha=0.3)
plt.show()
```

```python
# imported vs local cases and shows how the pandemic changed from imported cases to community spread
plt.figure(figsize=(14,6))

plt.plot(
    df["Date"],
    df["Daily Imported"],
    label="Imported"
)

plt.plot(
    df["Date"],
    df["Daily Local transmission"],
    label="Local"
)

plt.title("Imported vs Local COVID Cases")

plt.legend()
plt.grid(alpha=0.3)

plt.show()
```
